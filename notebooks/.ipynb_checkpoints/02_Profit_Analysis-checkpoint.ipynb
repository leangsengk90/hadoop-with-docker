# 02 — Profit Analysis
Profitability deep-dive: margin by category, sub-category, region, and loss-making orders.


```python
from pyhive import hive
import pandas as pd
import matplotlib.pyplot as plt

def get_conn():
    return hive.connect(host="hive-server2", port=10000, database="default", auth="NONE")
def fetch_df(cur, sql):
    cur.execute(sql)
    cols = [d[0].split(".")[-1] for d in cur.description]
    return pd.DataFrame(cur.fetchall(), columns=cols)

conn = get_conn(); cur = conn.cursor()
print("Connected.")
```

```python
# ── Profit Margin by Category ─────────────────────────────────────────────────
cat_profit = fetch_df(cur, """
    SELECT p.category,
           ROUND(SUM(oi.sales),2)   AS revenue,
           ROUND(SUM(oi.profit),2)  AS profit,
           ROUND(SUM(oi.profit)/SUM(oi.sales)*100,2) AS margin_pct
    FROM order_items oi JOIN products p ON oi.product_id = p.product_id
    GROUP BY p.category ORDER BY profit DESC
""")
print("=== Profit by Category ==="); display(cat_profit)

colors = ["green" if x >= 0 else "red" for x in cat_profit["profit"]]
fig, ax = plt.subplots(figsize=(7,4))
ax.barh(cat_profit["category"], cat_profit["profit"], color=colors)
ax.set_title("Total Profit by Category")
ax.set_xlabel("Profit (USD)")
plt.tight_layout(); plt.show()
```

```python
# ── Profit by Sub-Category (top 10 best & worst) ────────────────────────────
sub = fetch_df(cur, """
    SELECT p.sub_category,
           ROUND(SUM(oi.profit),2) AS profit,
           ROUND(SUM(oi.profit)/SUM(oi.sales)*100,2) AS margin_pct
    FROM order_items oi JOIN products p ON oi.product_id = p.product_id
    GROUP BY p.sub_category ORDER BY profit DESC
""")
display(sub)

colors = ["green" if x >= 0 else "red" for x in sub["profit"]]
fig, ax = plt.subplots(figsize=(10,6))
ax.barh(sub["sub_category"], sub["profit"], color=colors)
ax.axvline(0, color="black", linewidth=0.8)
ax.set_title("Profit by Sub-Category")
ax.set_xlabel("Profit (USD)")
plt.tight_layout(); plt.show()
```

```python
# ── Loss-Making Orders ───────────────────────────────────────────────────────
losses = fetch_df(cur, """
    SELECT oi.order_id, p.category, p.sub_category,
           ROUND(oi.sales,2) AS sales,
           ROUND(oi.profit,2) AS profit,
           ROUND(oi.discount*100,1) AS discount_pct
    FROM order_items oi JOIN products p ON oi.product_id = p.product_id
    WHERE oi.profit < 0
    ORDER BY oi.profit ASC
    LIMIT 20
""")
print(f"Showing top 20 most unprofitable line items:")
display(losses)

cur.close(); conn.close()
print("Done.")
```
