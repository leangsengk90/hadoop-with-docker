```python
import findspark
findspark.init()

from pyspark.sql import SparkSession

# Initialize Spark Session
spark = SparkSession.builder \
    .appName("Jupyter Spark Test") \
    .master("spark://namenode:7077") \
    .getOrCreate()

# Test Spark
data = [("Alice", 1), ("Bob", 2)]
columns = ["name", "id"]
df = spark.createDataFrame(data, columns)
df.show()

# Stop Spark
spark.stop()
```
