# Query Hive Tables via PyHive

Connects directly to **HiveServer2** (`hive-server2:10000`) using the `pyhive` library.

> **Why not `enableHiveSupport()`?**  
> The Spark binary installed (`spark-3.5.1-bin-hadoop3`) does not include Hive assembly JARs, so `enableHiveSupport()` fails with `'JavaPackage' object is not callable`.  
> PyHive is a pure-Python Hive client that works without any extra JARs.


## 1. Install Required Libraries


```python
import subprocess, sys

subprocess.check_call([
    sys.executable, "-m", "pip", "install", "--quiet",
    "pyhive[hive]", "thrift", "thrift-sasl", "sasl", "pandas"
])
print("Libraries installed.")
```

## 2. Connect to HiveServer2


```python
from pyhive import hive
import pandas as pd

conn = hive.connect(
    host="hive-server2",
    port=10000,
    database="default",
    auth="NONE",
)
cursor = conn.cursor()
print("Connected to HiveServer2 successfully!")
```

## 3. List Databases and Tables


```python
cursor.execute("SHOW DATABASES")
print("=== Databases ===")
for row in cursor.fetchall():
    print(" -", row[0])

cursor.execute("SHOW TABLES")
print("\n=== Tables in 'default' ===")
for row in cursor.fetchall():
    print(" -", row[1])
```

```python
## 4. Query `customers` Table
```

df_customers = pd.read_sql("SELECT * FROM default.customers", conn)

## 5. Filter Customers by Segment


```python
segment_filter = "Consumer"  # change to "Corporate" or "Home Office"

df_segment = spark.sql(f"""
    SELECT customer_id, customer_name, segment
    FROM default.customers
    WHERE segment = '{segment_filter}'
    ORDER BY customer_name
""")

print(f"Customers in segment '{segment_filter}': {df_segment.count()}")
df_segment.show(20, truncate=False)
```

## 6. Query `products` Table


```python
df_products = spark.sql("SELECT * FROM default.products")
print(f"Total rows: {df_products.count()}")
df_products.show(20, truncate=False)
```

## 7. Filter Products by Category


```python
category_filter = "Furniture"  # change to "Technology" or "Office Supplies"

df_cat = spark.sql(f"""
    SELECT *
    FROM default.products
    WHERE category = '{category_filter}'
    ORDER BY product_name
""")

print(f"Products in category '{category_filter}': {df_cat.count()}")
df_cat.show(20, truncate=False)
```

```python
category_filter = "Furniture"  # change to "Technology" or "Office Supplies"

df_cat = pd.read_sql(
    f"SELECT * FROM default.products WHERE category = '{category_filter}' ORDER BY product_name",
    conn
)
print(f"Products in category '{category_filter}': {len(df_cat)}")
display(df_cat)
```
