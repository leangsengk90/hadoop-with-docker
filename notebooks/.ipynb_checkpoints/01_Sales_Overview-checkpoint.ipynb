# 01 — Sales Overview
Key sales KPIs across the entire dataset: total revenue, orders, average order value, and monthly trends.


```python
from pyhive import hive
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

def get_conn():
    return hive.connect(host="hive-server2", port=10000, database="default", auth="NONE")

def fetch_df(cur, sql):
    cur.execute(sql)
    cols = [d[0].split(".")[-1] for d in cur.description]
    return pd.DataFrame(cur.fetchall(), columns=cols)

conn = get_conn()
cur = conn.cursor()
print("Connected.")
```

```python
# ── KPI Summary ──────────────────────────────────────────────────────────────
kpi = fetch_df(cur, """
    SELECT
        COUNT(DISTINCT o.order_id)           AS total_orders,
        COUNT(DISTINCT o.customer_id)        AS total_customers,
        ROUND(SUM(oi.sales), 2)              AS total_revenue,
        ROUND(AVG(oi.sales), 2)              AS avg_item_sales,
        ROUND(SUM(oi.profit), 2)             AS total_profit,
        ROUND(SUM(oi.profit)/SUM(oi.sales)*100, 2) AS profit_margin_pct
    FROM orders o
    JOIN order_items oi ON o.order_id = oi.order_id
""")
display(kpi)
```

```python
# ── Monthly Sales Trend ───────────────────────────────────────────────────────
monthly = fetch_df(cur, """
    SELECT SUBSTR(o.order_date,1,7) AS month,
           ROUND(SUM(oi.sales),2)   AS revenue,
           ROUND(SUM(oi.profit),2)  AS profit
    FROM orders o
    JOIN order_items oi ON o.order_id = oi.order_id
    GROUP BY SUBSTR(o.order_date,1,7)
    ORDER BY month
""")

fig, ax = plt.subplots(figsize=(14,4))
ax.plot(monthly["month"], monthly["revenue"], marker="o", label="Revenue")
ax.plot(monthly["month"], monthly["profit"],  marker="s", label="Profit")
ax.set_title("Monthly Revenue vs Profit")
ax.set_xlabel("Month")
ax.set_ylabel("USD")
ax.legend()
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# ── Annual Sales by Year ──────────────────────────────────────────────────────
yearly = fetch_df(cur, """
    SELECT SUBSTR(o.order_date,1,4) AS year,
           ROUND(SUM(oi.sales),2)   AS revenue,
           COUNT(DISTINCT o.order_id) AS orders
    FROM orders o
    JOIN order_items oi ON o.order_id = oi.order_id
    GROUP BY SUBSTR(o.order_date,1,4)
    ORDER BY year
""")

fig, ax1 = plt.subplots(figsize=(7,4))
ax2 = ax1.twinx()
ax1.bar(yearly["year"], yearly["revenue"], color="steelblue", label="Revenue")
ax2.plot(yearly["year"], yearly["orders"], color="orange", marker="o", label="Orders")
ax1.set_title("Annual Revenue & Order Volume")
ax1.set_ylabel("Revenue (USD)")
ax2.set_ylabel("# Orders")
fig.legend(loc="upper left", bbox_to_anchor=(0.1,0.9))
plt.tight_layout()
plt.show()
display(yearly)

cur.close(); conn.close()
print("Done.")
```
