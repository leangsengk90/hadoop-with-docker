```python
import findspark
findspark.init()
from pyspark import SparkContext

sc = SparkContext.getOrCreate()

data = "It is not over until it is over and until I win!"

def split_data(lines, separator):
    return lines.split(separator)

# Convert string into RDD
rdd = sc.parallelize([data])

new_rdd = (rdd
    .map(lambda x: split_data(x, "\n"))
    .flatMap(lambda x: x)
    .map(lambda x: split_data(x, " "))
    .flatMap(lambda x: x)
)
print(new_rdd)

word_count_map_reduce = (new_rdd
    .map(lambda x: (x, 1))
    .reduceByKey(lambda prev, new_value: prev+new_value)
)
print(word_count_map_reduce.collect())
print(word_count_map_reduce.take(3))

word_count_rdd = (new_rdd
    .groupBy(lambda word: word)
    .mapValues (lambda values: len(values))
)
print(word_count_rdd.take(5))
```
