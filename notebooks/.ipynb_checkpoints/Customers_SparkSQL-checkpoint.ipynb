# Customers - Spark SQL
Query customers data from HDFS using Spark SQL.

Update `HDFS_BASE` if your upload path differs.

```python
from pyspark.sql import SparkSession

# Start Spark (adjust master if needed)
spark = (SparkSession.builder
    .appName("Customers Spark SQL")
    .master("spark://namenode:7077")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("WARN")

# HDFS path where you uploaded the CSVs
HDFS_BASE = "hdfs:///user/hadoop/assessment-2"
CUSTOMERS_CSV = f"{HDFS_BASE}/customers.csv"

customers = (spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(CUSTOMERS_CSV)
    .withColumnRenamed("Customer ID", "customer_id")
    .withColumnRenamed("Customer Name", "customer_name")
    .withColumnRenamed("Segment", "segment")
)

customers.createOrReplaceTempView("customers")
print("Customers table ready.")
```

```python
# Spark SQL query
spark.sql("""
SELECT
  customer_id,
  customer_name,
  segment
FROM customers
ORDER BY customer_name
""").show(20, truncate=False)

# Optional: segment counts
spark.sql("""
SELECT segment, COUNT(*) AS customer_count
FROM customers
GROUP BY segment
ORDER BY customer_count DESC
""").show(truncate=False)
```
