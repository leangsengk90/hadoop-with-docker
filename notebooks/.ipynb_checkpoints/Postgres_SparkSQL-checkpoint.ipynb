# Postgres Direct Query via Spark SQL

This notebook connects directly to the `hive-postgres` container using Spark JDBC, then registers each table as a **Spark SQL temporary view** so you can query with plain SQL.

| Setting | Value |
|---|---|
| Container | `hive-postgres` |
| Database | `metastore` |
| User | `hive` |
| Password | `hive` |

## Step 1 — Discover Container IP

Port `5432` is not published to the host in `docker-compose.yml`, so we resolve the container's internal Docker IP at runtime.

```python
import subprocess

result = subprocess.run(
    [
        'docker', 'inspect', 'hive-postgres',
        '--format', '{{range .NetworkSettings.Networks}}{{.IPAddress}}{{end}}'
    ],
    capture_output=True, text=True
)

POSTGRES_IP = result.stdout.strip()
if not POSTGRES_IP:
    raise RuntimeError(
        "Could not resolve hive-postgres IP. "
        "Check the container is running: docker ps | grep hive-postgres"
    )

JDBC_URL = f"jdbc:postgresql://{POSTGRES_IP}:5432/metastore"
print(f"hive-postgres IP : {POSTGRES_IP}")
print(f"JDBC URL         : {JDBC_URL}")
```

## Step 2 — Create Spark Session

The `spark.jars.packages` config tells Spark to auto-download the Postgres JDBC driver via Maven.

```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("PostgresDirectSparkSQL")
    .config("spark.jars.packages", "org.postgresql:postgresql:42.7.3")
    .config("spark.sql.repl.eagerEval.enabled", True)   # pretty-print DataFrames
    .getOrCreate()
)
spark.sparkContext.setLogLevel("WARN")

print(f"Spark {spark.version} ready.")
```

## Step 3 — Connection Properties

```python
PG_PROPS = {
    "user":     "hive",
    "password": "hive",
    "driver":   "org.postgresql.Driver"
}

def read_pg(table: str):
    """Read a Postgres table into a Spark DataFrame."""
    return spark.read.jdbc(url=JDBC_URL, table=table, properties=PG_PROPS)

print("Helper function read_pg() defined.")
```

## Step 4 — Load Hive Metastore Tables as Spark SQL Views

We load the core Hive Metastore tables and register each as a temp view so Spark SQL can join them freely.

```python
# Hive Metastore uses UPPER-CASE quoted table names in Postgres
meta_tables = {
    "DBS":        'dbs',
    'TBLS':       'tbls',
    'COLUMNS_V2': 'columns',
    'SDS':        'sds',
    'TABLE_PARAMS': 'table_params',
}

for pg_table, view_name in meta_tables.items():
    df = read_pg(f'"{pg_table}"')
    df.createOrReplaceTempView(view_name)
    print(f'  Loaded "{pg_table}" -> view `{view_name}` ({df.count()} rows)')

print("\nAll views registered. Ready for Spark SQL.")
```

## Query 1 — List All Hive Databases

```python
spark.sql("""
    SELECT
        DB_ID,
        NAME        AS database_name,
        DESC        AS description,
        OWNER_NAME  AS owner
    FROM dbs
    ORDER BY DB_ID
""").show(truncate=False)
```

## Query 2 — List All Hive Tables

```python
spark.sql("""
    SELECT
        d.NAME      AS database_name,
        t.TBL_NAME  AS table_name,
        t.OWNER     AS owner,
        t.TBL_TYPE  AS table_type
    FROM tbls t
    JOIN dbs  d ON t.DB_ID = d.DB_ID
    ORDER BY d.NAME, t.TBL_NAME
""").show(truncate=False)
```

## Query 3 — List All Columns per Table

```python
spark.sql("""
    SELECT
        d.NAME          AS database_name,
        t.TBL_NAME      AS table_name,
        c.COLUMN_NAME,
        c.TYPE_NAME     AS data_type,
        c.INTEGER_IDX   AS col_order
    FROM columns c
    JOIN sds     s ON c.CD_ID = s.CD_ID
    JOIN tbls    t ON s.SD_ID = t.SD_ID
    JOIN dbs     d ON t.DB_ID = d.DB_ID
    ORDER BY d.NAME, t.TBL_NAME, c.INTEGER_IDX
""").show(50, truncate=False)
```

## Query 4 — Table HDFS Storage Locations

```python
spark.sql("""
    SELECT
        d.NAME          AS database_name,
        t.TBL_NAME      AS table_name,
        s.LOCATION      AS hdfs_location,
        s.INPUT_FORMAT
    FROM sds  s
    JOIN tbls t ON s.SD_ID = t.SD_ID
    JOIN dbs  d ON t.DB_ID = d.DB_ID
    ORDER BY d.NAME, t.TBL_NAME
""").show(truncate=False)
```

## Query 5 — Table Parameters (Properties)

This shows properties like `skip.header.line.count`, `transient_lastDdlTime`, etc.

```python
spark.sql("""
    SELECT
        t.TBL_NAME  AS table_name,
        p.PARAM_KEY,
        p.PARAM_VALUE
    FROM table_params p
    JOIN tbls        t ON p.TBL_ID = t.TBL_ID
    ORDER BY t.TBL_NAME, p.PARAM_KEY
""").show(50, truncate=False)
```

## Custom SQL Query

Use the registered views — `dbs`, `tbls`, `columns`, `sds`, `table_params` — to write any query you like.

```python
# ---- Write your own Spark SQL here ----
spark.sql("""
    SELECT
        d.NAME     AS database_name,
        COUNT(*)   AS table_count
    FROM tbls t
    JOIN dbs  d ON t.DB_ID = d.DB_ID
    GROUP BY d.NAME
    ORDER BY table_count DESC
""").show()
```

## Stop Spark Session

```python
spark.stop()
print("Spark session stopped.")
```
