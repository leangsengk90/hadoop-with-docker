# Spark RDD Word Count

This notebook demonstrates how to use the Apache Spark **RDD (Resilient Distributed Dataset) API** to perform a word count on a text file.

### Steps Involved:
1. **Initialize Spark:** Setting up the `SparkContext` using `findspark`.
2. **Load Data:** Reading a text file from the local filesystem.
3. **Process:** Using `flatMap`, `map`, and `reduceByKey` to count occurrences.
4. **Output:** Sorting the counts and collecting all results for display.

```python
import findspark
findspark.init()

from pyspark import SparkContext

# Get or create a SparkContext
# Using getOrCreate avoids errors if the context already exists in the kernel
sc = SparkContext.getOrCreate(appName="Full_WordCount_RDD")

print(f"Spark initialized. App: {sc.appName}")
```

### Execute Word Count

The following cell processes the text file. We use the `file://` prefix to specify the absolute path within the container.

```python
# Define input path
input_path = "file:///usr/local/hadoop/etc/hadoop/homework-sparkrdd/BattleCreekDec19_2019.txt"

try:
    # RDD Workflow:
    # 1. Read lines into RDD
    # 2. Split lines into words (flatMap)
    # 3. Clean words (lowercase, strip whitespace, remove empty strings)
    # 4. Map each word to (word, 1)
    # 5. Aggregate counts (reduceByKey)
    # 6. Sort results by frequency (sortBy)
    
    word_counts = sc.textFile(input_path) \
                    .flatMap(lambda line: line.split(" ")) \
                    .map(lambda word: (word.lower().strip().strip('.,!?:;()'), 1)) \
                    .filter(lambda x: x[0] != "") \
                    .reduceByKey(lambda a, b: a + b) \
                    .sortBy(lambda x: -x[1])

    # Use .collect() to retrieve ALL results back to the notebook
    all_results = word_counts.collect()

    print(f"Success! Total {len(all_results)} unique words found.\n")
    
    # Print all occurrences
    for word, count in all_results:
        print(f"'{word}': {count}")

except Exception as e:
    print(f"Error processing file: {e}")
```
